Implementación del modelo de Xu y Tenenbaum (2007): las hipótesis son
nodos de un árbol taxonómico, el prior es la **longitud de rama** (*branch length*) y la
verosimilitud es el principio del tamaño.

La ecuación que vas a usar, para $h_1 \subset h_2$:

$$\frac{P(h_2 \mid X)}{P(h_1 \mid X)} = r\,s^{-n},
\qquad r=\frac{P(h_2)}{P(h_1)},\quad s=\frac{|h_2|}{|h_1|},
\qquad n^{*}=\frac{\ln r}{\ln s}$$

```python
import numpy as np, matplotlib.pyplot as plt

# --- el árbol: 3 dálmatas, 3 otros perros, 3 otros animales ---
OBJETOS = ["dálmata1","dálmata2","dálmata3","terrier","labrador","basset",
           "gato","tucán","foca"]

ARBOL = {                      # nodo: (extensión, longitud de rama = prior no normalizado)
    "h_sub  (dálmata)":  (set(range(0, 3)), 0.55),
    "h_bas  (perro)":    (set(range(0, 6)), 1.90),
    "h_sup  (animal)":   (set(range(0, 9)), 0.50),
}

def posterior(X, arbol=ARBOL, fuerte=True):
    """X es una lista de índices de objetos etiquetados."""
    post = {}
    for nombre, (ext, rama) in arbol.items():
        if not set(X) <= ext: continue
        post[nombre] = rama * ((1/len(ext))**len(X) if fuerte else 1.0)
    Z = sum(post.values())
    return {k: v/Z for k, v in post.items()}

def generaliza(X, arbol=ARBOL, fuerte=True):
    """P(objeto y pertenece al concepto | X), para cada objeto."""
    post = posterior(X, arbol, fuerte)
    g = np.zeros(len(OBJETOS))
    for nombre, p in post.items():
        for y in arbol[nombre][0]: g[y] += p
    return g, post

def mostrar(X, arbol=ARBOL, fuerte=True):
    g, post = generaliza(X, arbol, fuerte)
    print(f"X = {[OBJETOS[i] for i in X]}   ({'teacher-driven' if fuerte else 'learner-driven'})")
    for k, v in post.items(): print(f"   P({k}) = {v:.3f}")
    plt.figure(figsize=(7, 1.8))
    plt.bar(OBJETOS, g, color=["crimson" if i in X else "navy" for i in range(9)])
    plt.ylim(0, 1.05); plt.ylabel("P(y ∈ C | X)"); plt.xticks(rotation=35, ha="right")
    plt.show()

mostrar([0])          # un ejemplo
mostrar([0, 1, 2])    # tres ejemplos subordinados
```
